```python
import matplotlib.pyplot as plt
import seaborn as sns
```

### Plot the number of films that have been released each decade over the history of cinema.

```python
films_per_decade= cast.groupby((cast['year']//10)*10)['title'].count().reset_index()
print(films_per_decade)
```

```python
sns.barplot(data= films_per_decade, x = 'year', y = 'title')
plt.xticks(rotation=45)
```

### Plot the number of "Hamlet" films made each decade.

```python
hamlet_per_decade = (cast[cast['title']== 'Hamlet'].groupby((cast['year']//10)*10)['title'].count().reset_index())
print(hamlet_per_decade)
```

```python
ax= sns.barplot(data=hamlet_per_decade, x = 'year', y = 'title', palette ='plasma',saturation=1)
plt.figure(figsize=(12,5))
for container in ax.containers:       # this code is for show data labels
    ax.bar_label(container)
```

### Plot the number of "Rustler" characters in each decade of the history of film.

```python
Rustler_per_decade = (cast[cast['character']== 'Rustler'].groupby((cast['year']//10)*10)['character'].count().reset_index())
print(Rustler_per_decade)
```

```python
ax= sns.barplot(data=Rustler_per_decade, x = 'year', y = 'character', palette ='magma',saturation=1)
plt.figure(figsize=(12,5))
for container in ax.containers:       # this code is for show data labels
    ax.bar_label(container)
```

### Plot the number of "Hamlet" characters each decade.

```python
hamlet_char_per_decade = (cast[cast['title']== 'Hamlet'].groupby((cast['year']//10)*10)['character'].nunique().reset_index())
print(hamlet_char_per_decade)
```

```python
ax= sns.barplot(data=hamlet_char_per_decade, x = 'year', y = 'character', palette ='inferno',saturation=1)
plt.figure(figsize=(12,5))
for container in ax.containers:       # this code is for show data labels
    ax.bar_label(container)
```
